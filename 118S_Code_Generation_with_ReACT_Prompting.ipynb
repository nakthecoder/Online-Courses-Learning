```python
# 🧠 Personal Finance Customer Service Flow (ReACT-style Simulation)

def analyze_expenses():
    return "Your average monthly spending is $3,850, with dining and shopping being your top categories."

def suggest_savings_plan():
    return "You could save about $250 per month by reducing dining and shopping by 15% each."

def forecast_budget():
    return "If you follow this savings plan, you’ll save around $3,000 over the next 12 months."

def schedule_coaching_session():
    return "I've scheduled a 30-minute financial coaching session for you this Friday at 3 PM."

def budget_health_check():
    return "Your budget health score is 78/100 — stable, but improving savings consistency could boost it above 85."

def ask_satisfaction():
    return "Are you satisfied with this assistance, or would you like to adjust your savings plan?"

# Main customer service workflow
def finance_service_flow():
    print("👋 Welcome to FinAssist — your personal finance advisor.\n")
    print("How can I help you today?")
    print("1. View expense summary")
    print("2. Get savings suggestions")
    print("3. Forecast yearly budget")
    print("4. Schedule a financial coaching session")
    print("5. Instant budget health check\n")

    choice = input("Enter a number (1–5): ").strip()

    if choice == "1":
        print("\nAction: Analyzing expenses...")
        print("Observation:", analyze_expenses())
    elif choice == "2":
        print("\nAction: Suggesting savings plan...")
        print("Observation:", suggest_savings_plan())
    elif choice == "3":
        print("\nAction: Forecasting your budget...")
        print("Observation:", forecast_budget())
    elif choice == "4":
        print("\nAction: Scheduling a coaching session...")
        print("Observation:", schedule_coaching_session())
    elif choice == "5":
        print("\nAction: Performing instant budget health check...")
        print("Observation:", budget_health_check())
    else:
        print("\nSorry, I didn’t understand that option.")

    print("\n" + ask_satisfaction())

# Run the flow
if __name__ == "__main__":
    finance_service_flow()
```
