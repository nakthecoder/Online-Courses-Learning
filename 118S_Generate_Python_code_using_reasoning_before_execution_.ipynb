```python
# 🧠 Research Paper Self-Reflection Summarizer
# Simulates an AI summarizing, critiquing, and improving its own output.

def generate_initial_summary(paper_title):
    """Simulate the AI's first attempt at summarizing a research paper."""
    return (
        f"The paper '{paper_title}' discusses deep learning methods for climate prediction. "
        "It uses a neural network to analyze temperature data and claims high accuracy in forecasting. "
        "The study concludes that AI can significantly improve climate modeling."
    )

def self_critique(summary):
    """Simulate a self-reflection critique of the AI's own summary."""
    critique = (
        "The summary captures the general topic and conclusion but lacks details about the dataset, "
        "evaluation methods, and limitations. It also uses vague terms like 'high accuracy' without metrics. "
        "The explanation could be more precise and balanced, avoiding overstatements."
    )
    return critique

def improve_summary(summary, critique):
    """Simulate the AI revising its summary based on its own critique."""
    improved = (
        "The paper explores how deep learning can enhance climate prediction by using a neural network "
        "trained on 50 years of global temperature and atmospheric data. "
        "The authors compare their model with traditional forecasting methods and report a 12% improvement "
        "in predictive accuracy. They also note that data scarcity and regional variability remain key challenges."
    )
    return improved

def summarize_with_self_reflection(paper_title):
    print("🧩 Step 1: Initial Summary\n")
    initial = generate_initial_summary(paper_title)
    print(initial, "\n")

    print("🔍 Step 2: Self-Critique\n")
    critique = self_critique(initial)
    print(critique, "\n")

    print("✅ Step 3: Improved Summary\n")
    improved = improve_summary(initial, critique)
    print(improved, "\n")

# Example run
if __name__ == "__main__":
    summarize_with_self_reflection("Deep Learning for Climate Forecasting")
```
